## Inference Data & Result

I will create different inference data to test the model accuracy. I will give suggestion based on the results

### 1. Fresh graduate

```python
freshgrad = {
    'Age': [22, 23, 22], 
    'DailyRate': [1000,400,400], # One candidate has referrals while others don't yet
    'DistanceFromHome': [10,20,5], 
    'MonthlyIncome': [2000, 2000, 2000], #Assuming same salary
    'TotalWorkingYears': [0,0,0], 
    'TrainingTimesLastYear': [0,0,0], 
    'YearsAtCompany': [0,0,0], 
    'BusinessTravel':['Travel_Frequently', 'Travel_Rarely', 'Non-Travel'], 
    'Department': ['Sales', 'Human Resources', 'Research & Development'], # Adjusted based on the business travel
    'OverTime': ['Yes', 'Yes', 'Yes'], #All newly hires usually does overtime to leave a good impression for the company 
    'EnvironmentSatisfaction': [3,2,2], 
    'JobSatisfaction': [2,2,2], 
    'StockOptionLevel': [0,0,0], 
    'WorkLifeBalance': [1,1,1]
}
freshgrad = pd.DataFrame(freshgrad)
freshgrad
```

```python
model_best
```

```python
y_pred_inf = model_best.predict(freshgrad)
print(y_pred_inf)
```

According to this machine, fresh graduate that hasn't worked will have higher chance of attrition. Attrition is usually caused by:
1.  Inadequate compensation/benefit
2.  Work/life imbalance
3.  Unrealistic Expectation
4.  Ineffective management practive
5.  Misallignment with workplace culture
6.  Lack of career advancement opportunities

For fresh graduate, these might be the cause for them to get attrited. So company must do the opposite to keep these type of people.

https://www.shrm.org/topics-tools/news/employee-relations/attrition-definition-types-causes-mitigation-tips

### 5. People in retirement age

```python
old = {
    'Age': [70, 73, 72], 
    'DailyRate': [50,50,100],
    'DistanceFromHome': [10,20,5], 
    'MonthlyIncome': [10000, 9000, 8500], 
    'TotalWorkingYears': [45,46,47], 
    'TrainingTimesLastYear': [0,0,0], 
    'YearsAtCompany': [45,46,47], 
    'BusinessTravel':['Travel_Frequently', 'Travel_Rarely', 'Non-Travel'], 
    'Department': ['Sales', 'Human Resources', 'Research & Development'], 
    'OverTime': ['No', 'No', 'No'], #Old people cannot have overtime
    'EnvironmentSatisfaction': [2,2,2], 
    'JobSatisfaction': [2,2,2], 
    'StockOptionLevel': [1,1,1], 
    'WorkLifeBalance': [3,3,3]
}
old= pd.DataFrame(old)
y_pred_old = model_best.predict(old)
print(y_pred_old)
```

These are not the result we want because these are the age of retirement. This is an example of false negative in our model

## Insight
- From the freshgraduate data, we can see that attrition may happen because the model was made for employees within the company. Employee outside will have higher chance of being marked as attrited because this model was not made to analyze potential candidates
- This model still has the tendency to have false negative prediction from what we see in the data for people in retirement age, which is what we want to minimize as much as possible.
